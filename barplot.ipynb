```python
import matplotlib.pyplot as plt

import numpy as np
```

```python
# Department salary data
 
departments = ['Engineering', 'Marketing', 
               'Sales', 'HR', 'Finance'] 
avg_salary  = [95000, 75000, 70000, 
               65000, 85000] 
max_salary  = [150000, 110000, 120000, 
               90000, 130000] 

# make the subplots
fig, ax = plt.subplots(1, 2, figsize=(12, 6))

x = np.arange(len(departments)) 
width = 0.35     # bar width 

# --- Subplot 1: Average Salary ---
ax[0].bar(x - width/2, 
          avg_salary, 
          width, 
          label='Average Salary', 
          color='purple', 
          edgecolor='black', 
          linewidth=0.5)

# Formatting subplot 0
ax[0].set_title('Average Salary by Department')
ax[0].set_xticks(x)
ax[0].set_xticklabels(departments, rotation=15)
ax[0].set_ylabel('Salary ($)')
ax[0].legend()
# Note: adjusted ylim automatically so Engineering (95k) isn't clipped
ax[0].set_ylim(0, 110000) 

# --- Subplot 2: Maximum Salary ---
ax[1].bar(x + width/2, 
          max_salary, # Changed to max_salary to visualize the full dataset
          width, 
          label='Maximum Salary', 
          color='lightgreen', 
          edgecolor='black',
          linewidth=0.5)

# Formatting subplot 1
ax[1].set_title('Maximum Salary by Department')
ax[1].set_xticks(x)
ax[1].set_xticklabels(departments, rotation=15)
ax[1].set_ylabel('Salary ($)')
ax[1].legend()
ax[1].set_ylim(0, 170000)

plt.tight_layout()
plt.show()


```
