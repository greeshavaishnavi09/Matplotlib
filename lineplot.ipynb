```python
import matplotlib.pyplot as plt

import numpy as np
```

```python

# create the data

months = ['Jan','Feb','Mar','Apr','May','Jun', 
          'Jul','Aug','Sep','Oct','Nov','Dec'] 

product_a = [120, 135, 148, 162, 155, 178, 
             195, 210, 198, 220, 235, 250] 

product_b = [80,  95,  88, 105, 112, 120, 
             135, 128, 142, 155, 160, 175] 

product_c = [60,  65,  70,  75,  85,  90, 
             95,  100, 108, 115, 120, 130]

# make the plot 
X = range(len(months))  # range of months column

# make the subplots
fig, ax = plt.subplots(2,2,figsize=(12, 6))

# make the plot

ax[0,0].plot(X,product_a,
       color = 'purple',
       linewidth = 2,
       marker = "o",
       markersize = 6,
       label = "Product A")
ax[0,0].set_title("Product A")
ax[0,0].legend(["Product A"])


ax[0,1].plot(X,product_b,
       color = 'lightgreen',
       linewidth = 2,
       marker = "o",
       markersize = 6,
       label = "Product B")
ax[0,1].set_title("Product B")
ax[0,1].legend(["Product B"])

 

ax[1,0].plot(X,product_c,
       color = 'lightblue',
       linewidth = 2,
       marker = "o",
       markersize = 6,
       label = "Product C")
ax[1,0].set_title("Product C")
ax[1,0].legend(["Product C"])

 

ax[1,1].plot(product_a,product_b,
       color = 'brown',
       linewidth = 2,
       marker = "o",
       markersize = 6,
       label = "Product A"
       )
ax[1,1].set_title("Product B vs A")
ax[1,1].legend(["Product B vs A"])

plt.show()


```
