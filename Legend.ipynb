```python
import matplotlib.pyplot as plt
import numpy as np

# create the plot

# create the data
X = np.linspace(0, 15, 200)
y = np.sin(X)

# create the figure
plt.figure(figsize=(10, 8))

# plot the data
plt.plot(
    X,
    y,
    color="red",
    marker="*",
    markeredgecolor="black",
    markerfacecolor="orange",
    markersize=2,
)

# Added missing parentheses to execute the function
plt.show()
```

```python
# add the legend
plt.legend(["Sine Waves"],

          loc = "lower left",

          fontsize = 15,

          title = "Info",

          title_fontsize = 12,

          edgecolor='black',

          facecolor = "lightblue",

          shadow = True

          )
```
